```python
df = pd.read_csv(DATA_FILE, low_memory=False)

print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")
display(df.head())
print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Inspect structure and data quality
display(df.info())
display(df.describe(include="all").T.head(30))

quality = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "missing": df.isna().sum(),
    "missing_pct": df.isna().mean().mul(100).round(2),
    "unique": df.nunique(dropna=True)
}).sort_values("missing_pct", ascending=False)

display(quality)
quality.to_csv(TABLE_DIR / "data_quality.csv")
```

```python
# ============================================================
# 7. CLEAN COLUMN NAMES
# ============================================================

df.columns = [
    column.strip().lower()
    for column in df.columns
]


required_columns = {
    "year_month",
    "month_of_release",
    "passenger_type",
    "direction",
    "citizenship",
    "visa",
    "country_of_residence",
    "estimate",
    "standard_error",
    "status",
}


missing_columns = required_columns - set(df.columns)


if missing_columns:

    raise ValueError(
        f"Missing columns: {missing_columns}"
    )

```

```python
# ============================================================
# 8. CLEAN DATA TYPES
# ============================================================

df["year_month"] = pd.to_datetime(
    df["year_month"],
    errors="coerce"
)


df["estimate"] = pd.to_numeric(
    df["estimate"],
    errors="coerce"
)


df["standard_error"] = pd.to_numeric(
    df["standard_error"],
    errors="coerce"
)


text_columns = [
    "passenger_type",
    "direction",
    "citizenship",
    "visa",
    "country_of_residence",
    "status",
]


for column in text_columns:

    df[column] = (
        df[column]
        .astype(str)
        .str.strip()
    )


df = df.dropna(
    subset=[
        "year_month",
        "estimate"
    ]
).copy()


df["year"] = df["year_month"].dt.year

df["month"] = df["year_month"].dt.month

df["month_name"] = (
    df["year_month"]
    .dt.strftime("%b")
)
```

```python
# ============================================================
# 9. IDENTIFY TOTAL DIMENSIONS
# ============================================================

def has_total(column):

    values = set(
        df[column]
        .dropna()
        .astype(str)
    )

    return "TOTAL" in values
```

```python
# ============================================================
# 10. FILTER TO TOTAL SERIES
# ============================================================
#
# The raw dataset contains several dimensions.
#
# If TOTAL exists, use it to avoid counting the same migration
# flow repeatedly across visa/citizenship categories.
# ============================================================

analysis = df.copy()


if has_total("passenger_type"):

    analysis = analysis[
        analysis["passenger_type"] == "TOTAL"
    ]


if has_total("direction"):

    analysis = analysis[
        analysis["direction"] == "TOTAL"
    ]


if has_total("citizenship"):

    analysis = analysis[
        analysis["citizenship"] == "TOTAL"
    ]


if has_total("visa"):

    analysis = analysis[
        analysis["visa"] == "TOTAL"
    ]


print(
    "\nRows after TOTAL filtering:",
    f"{len(analysis):,}"
)
```

```python
# ============================================================
# 11. MONTHLY COUNTRY SERIES
# ============================================================

monthly_country = (
    analysis
    .groupby(
        [
            "year_month",
            "country_of_residence"
        ],
        as_index=False
    )["estimate"]
    .sum()
)


monthly_country.to_csv(
    TABLE_DIR /
    "monthly_country_series.csv",
    index=False
)
```

```python
# ============================================================
# 12. NIGERIA MONTHLY SERIES
# ============================================================

nigeria_monthly = monthly_country[
    monthly_country[
        "country_of_residence"
    ]
    .str.casefold()
    == "nigeria"
].copy()


nigeria_monthly = nigeria_monthly.sort_values(
    "year_month"
)


nigeria_monthly = nigeria_monthly.rename(
    columns={
        "estimate": "nigeria"
    }
)
```

```python
# ============================================================
# 13. COUNTRY GLOBAL RANKING
# ============================================================

country_totals = (
    analysis
    .groupby(
        "country_of_residence",
        as_index=False
    )["estimate"]
    .sum()
    .sort_values(
        "estimate",
        ascending=False
    )
)


country_totals["rank"] = np.arange(
    1,
    len(country_totals) + 1
)


country_totals.to_csv(
    TABLE_DIR /
    "global_country_ranking.csv",
    index=False
)


nigeria_global = country_totals[
    country_totals[
        "country_of_residence"
    ]
    .str.casefold()
    == "nigeria"
]


nigeria_global_rank = int(
    nigeria_global["rank"].iloc[0]
)
```

```python
# ============================================================
# 14. REGIONAL CLASSIFICATION
# ============================================================

analysis["region"] = np.where(
    analysis["country_of_residence"]
    .isin(WEST_AFRICA),
    "West Africa",
    np.where(
        analysis["country_of_residence"]
        .isin(AFRICA),
        "Africa",
        "Other"
    )
)
```

```python
# ============================================================
# 15. REGIONAL MONTHLY SERIES
# ============================================================

west_africa_monthly = (
    analysis[
        analysis["region"]
        == "West Africa"
    ]
    .groupby(
        "year_month",
        as_index=False
    )["estimate"]
    .sum()
)


west_africa_monthly = (
    west_africa_monthly
    .rename(
        columns={
            "estimate": "west_africa"
        }
    )
)


africa_monthly = (
    analysis[
        analysis["region"]
        == "Africa"
    ]
    .groupby(
        "year_month",
        as_index=False
    )["estimate"]
    .sum()
)


africa_monthly = (
    africa_monthly
    .rename(
        columns={
            "estimate": "africa"
        }
    )
)


world_monthly = (
    analysis
    .groupby(
        "year_month",
        as_index=False
    )["estimate"]
    .sum()
)


world_monthly = (
    world_monthly
    .rename(
        columns={
            "estimate": "world"
        }
    )
)
```

```python
# ============================================================
# 16. COMBINE REGIONAL SERIES
# ============================================================

regional_monthly = (
    nigeria_monthly[
        [
            "year_month",
            "nigeria"
        ]
    ]
    .merge(
        west_africa_monthly,
        on="year_month",
        how="outer"
    )
    .merge(
        africa_monthly,
        on="year_month",
        how="outer"
    )
    .merge(
        world_monthly,
        on="year_month",
        how="outer"
    )
    .sort_values("year_month")
)


regional_monthly.to_csv(
    TABLE_DIR /
    "monthly_regional_series.csv",
    index=False
)
```

```python
# ============================================================
# 17. ANNUAL SERIES
# ============================================================

regional_monthly["year"] = (
    regional_monthly[
        "year_month"
    ].dt.year
)


annual = (
    regional_monthly
    .groupby(
        "year",
        as_index=False
    )[
        [
            "nigeria",
            "west_africa",
            "africa",
            "world"
        ]
    ]
    .sum()
)


# Complete years only
annual_complete = annual[
    annual["year"] <= 2025
].copy()
```

```python
# ============================================================
# 18. CALCULATE NIGERIA SHARES
# ============================================================

annual_complete[
    "nigeria_share_west_africa_pct"
] = (
    annual_complete["nigeria"]
    /
    annual_complete["west_africa"]
    * 100
)
annual_complete[
    "nigeria_share_africa_pct"
] = (
    annual_complete["nigeria"]
    /
    annual_complete["africa"]
    * 100
)
annual_complete[
    "nigeria_share_world_pct"
] = (
    annual_complete["nigeria"]
    /
    annual_complete["world"]
    * 100
)
annual_complete.to_csv(
    TABLE_DIR /
    "annual_regional_series_2001_2025.csv",
    index=False
)
```

```python
# ============================================================
# 19. CUMULATIVE STATISTICS
# ============================================================
nigeria_total = (
    nigeria_monthly["nigeria"]
    .sum()
)
west_total = (
    west_africa_monthly["west_africa"]
    .sum()
)
africa_total = (
    africa_monthly["africa"]
    .sum()
)
world_total = (
    world_monthly["world"]
    .sum()
)
nigeria_share_west = (
    nigeria_total
    /
    west_total
    * 100
)
nigeria_share_africa = (
    nigeria_total
    /
    africa_total
    * 100
)
nigeria_share_world = (
    nigeria_total
    /
    world_total
    * 100
)
summary = pd.DataFrame(
    {
        "metric": [
            "Nigeria cumulative arrivals",
            "West Africa cumulative arrivals",
            "Africa cumulative arrivals",
            "World cumulative arrivals",
            "Nigeria share of West Africa (%)",
            "Nigeria share of Africa (%)",
            "Nigeria share of World (%)",
        ],
        "value": [
            nigeria_total,
            west_total,
            africa_total,
            world_total,
            nigeria_share_west,
            nigeria_share_africa,
            nigeria_share_world,
        ]
    }
)
summary.to_csv(
    TABLE_DIR /
    "summary_statistics.csv",
    index=False
)
```

```python
# ============================================================
# 20. WEST AFRICA RANKING
# ============================================================

west_africa_ranking = (
    analysis[
        analysis["country_of_residence"]
        .isin(WEST_AFRICA)
    ]
    .groupby(
        "country_of_residence",
        as_index=False
    )["estimate"]
    .sum()
    .sort_values(
        "estimate",
        ascending=False
    )
)
west_africa_ranking["rank"] = (
    np.arange(
        1,
        len(west_africa_ranking) + 1
    )
)
west_africa_ranking.to_csv(
    TABLE_DIR /
    "west_africa_ranking.csv",
    index=False
)
```

```python

# ============================================================
# 21. AFRICA RANKING
# ============================================================

africa_ranking = (
    analysis[
        analysis["country_of_residence"]
        .isin(AFRICA)
    ]
    .groupby(
        "country_of_residence",
        as_index=False
    )["estimate"]
    .sum()
    .sort_values(
        "estimate",
        ascending=False
    )
)
africa_ranking["rank"] = (
    np.arange(
        1,
        len(africa_ranking) + 1
    )
)
africa_ranking.to_csv(
    TABLE_DIR /
    "africa_ranking.csv",
    index=False
)
```

```python
# ============================================================
# 22. 2025 GLOBAL RANKING
# ============================================================

global_2025 = (
    analysis[
        analysis["year"] == 2025
    ]
    .groupby(
        "country_of_residence",
        as_index=False
    )["estimate"]
    .sum()
    .sort_values(
        "estimate",
        ascending=False
    )
)
global_2025["rank"] = (
    np.arange(
        1,
        len(global_2025) + 1
    )
)
global_2025.to_csv(
    TABLE_DIR /
    "global_2025_ranking.csv",
    index=False
)
```

```python
# ============================================================
# 23. NIGERIA AFRICA RANK
# ============================================================

nigeria_africa_row = africa_ranking[
    africa_ranking[
        "country_of_residence"
    ]
    .str.casefold()
    == "nigeria"
]
nigeria_africa_rank = int(
    nigeria_africa_row["rank"].iloc[0]
)
```

```python
# ============================================================
# 24. NIGERIA WEST AFRICA RANK
# ============================================================

nigeria_west_row = west_africa_ranking[
    west_africa_ranking[
        "country_of_residence"
    ]
    .str.casefold()
    == "nigeria"
]
nigeria_west_rank = int(
    nigeria_west_row["rank"].iloc[0]
)
```

```python
# 25. VISA ANALYSIS
# ============================================================

nigeria_raw = df[
    df["country_of_residence"]
    .str.casefold()
    == "nigeria"
].copy()


visa_nigeria = (
    nigeria_raw
    .groupby(
        "visa",
        as_index=False
    )["estimate"]
    .sum()
    .sort_values(
        "estimate",
        ascending=False
    )
)
# Remove TOTAL from the composition
visa_nigeria = visa_nigeria[
    visa_nigeria["visa"]
    .str.upper()
    != "TOTAL"
].copy()
visa_nigeria["share_pct"] = (
    visa_nigeria["estimate"]
    /
    visa_nigeria["estimate"].sum()
    * 100
)
visa_nigeria.to_csv(
    TABLE_DIR /
    "nigeria_visa_composition.csv",
    index=False
)
```

## CORRELATION MATRIX

##  CORRELATION SIGNIFICANCE TESTS

```python
# ============================================================
# 27. CORRELATION SIGNIFICANCE TESTS
# ============================================================

correlation_pairs = [
    (
        "Nigeria",
        "West Africa",
        "nigeria",
        "west_africa"
    ),
    (
        "Nigeria",
        "Africa",
        "nigeria",
        "africa"
    ),
    (
        "Nigeria",
        "World",
        "nigeria",
        "world"
    )
]
correlation_results = []

for name_x, name_y, x, y in correlation_pairs:

    r, p = stats.pearsonr(
        annual_complete[x],
        annual_complete[y]
    )

    correlation_results.append(
        {
            "variable_x": name_x,
            "variable_y": name_y,
            "pearson_r": r,
            "p_value": p,
            "n": len(annual_complete)
        }
    )
correlation_results = pd.DataFrame(
    correlation_results
)
correlation_results.to_csv(
    TABLE_DIR /
    "correlation_tests.csv",
    index=False
)

```

## OLS REGRESSION

```python
# ============================================================
# 28. OLS REGRESSION
# ============================================================
regression_data = annual_complete.copy()
regression_data["time"] = (
    regression_data["year"]
    -
    regression_data["year"].min()
)
X = sm.add_constant(
    regression_data["time"]
)
y = regression_data["nigeria"]

ols_model = sm.OLS(
    y,
    X
).fit()

print("\nOLS REGRESSION")

print(
    ols_model.summary()
)
with open(
    OUTPUT_DIR /
    "ols_regression_summary.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write(
        ols_model.summary().as_text()
    )
ols_coefficients = pd.DataFrame(
    {
        "term": ols_model.params.index,
        "coefficient": ols_model.params.values,
        "standard_error": ols_model.bse.values,
        "p_value": ols_model.pvalues.values,
    }
)
ols_coefficients.to_csv(
    TABLE_DIR /
    "ols_regression_coefficients.csv",
    index=False
)

```

## SEGMENTED REGRESSION post-2022 change

```python
# ============================================================
# 29. SEGMENTED REGRESSION
# ============================================================
#
# Intervention begins in 2023.
#
# post_2022:
#       0 before 2023
#       1 from 2023 onward
#
# post_2022_time:
#       0 before 2023
#       1 in 2023
#       2 in 2024
#       3 in 2025
# ============================================================

segmented = annual_complete.copy()


segmented["post_2022"] = (
    segmented["year"] >= 2023
).astype(int)
segmented["post_2022_time"] = np.where(
    segmented["year"] >= 2023,
    segmented["year"] - 2022,
    0
)

X_segmented = sm.add_constant(
    segmented[
        [
            "post_2022",
            "post_2022_time"
        ]
    ]
)
segmented_model = sm.OLS(
    segmented["nigeria"],
    X_segmented
).fit()

print("\nSEGMENTED REGRESSION")

print(
    segmented_model.summary()
)
with open(
    OUTPUT_DIR /
    "segmented_regression_summary.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write(
        segmented_model.summary().as_text()
    )
segmented_coefficients = pd.DataFrame(
    {
        "term": segmented_model.params.index,
        "coefficient": segmented_model.params.values,
        "standard_error": segmented_model.bse.values,
        "p_value": segmented_model.pvalues.values,
    }
)
segmented_coefficients.to_csv(
    TABLE_DIR /
    "segmented_regression_coefficients.csv",
    index=False
)
```

##  HYPOTHESIS TESTING,WELCH T-TEST,MANN-WHITNEY U TEST and CHI-SQUARE VISA COMPOSITION TEST

```python
# ============================================================
# 30. HYPOTHESIS TESTING
# ============================================================
#
# H0:
#     Nigeria's 2023–2025 annual migration level is not
#     statistically different from its 2001–2019 level.
#
# H1:
#     The two periods differ.
# ============================================================

pre_covid = annual_complete.loc[
    annual_complete["year"].between(
        2001,
        2019
    ),
    "nigeria"
]

recent = annual_complete.loc[
    annual_complete["year"].between(
        2023,
        2025
    ),
    "nigeria"
]

# ------------------------------------------------------------
# Welch t-test
# ------------------------------------------------------------

welch_t, welch_p = stats.ttest_ind(
    pre_covid,
    recent,
    equal_var=False
)

# ------------------------------------------------------------
# Mann-Whitney U test
# ------------------------------------------------------------

mann_whitney_u, mann_whitney_p = (
    stats.mannwhitneyu(
        pre_covid,
        recent,
        alternative="two-sided"
    )
)

# ============================================================
# 31. CHI-SQUARE VISA COMPOSITION TEST
# ============================================================
#
# Compare Nigeria against the rest of West Africa,
# 2023–2025.
# ============================================================
visa_comparison = df[
    df["year"].between(
        2023,
        2025
    )
].copy()

visa_comparison["group"] = np.where(
    visa_comparison[
        "country_of_residence"
    ]
    .str.casefold()
    == "nigeria",
    "Nigeria",
    np.where(
        visa_comparison[
            "country_of_residence"
        ].isin(WEST_AFRICA),
        "Rest of West Africa",
        "Other"
    )
)

visa_comparison = visa_comparison[
    visa_comparison["group"].isin(
        [
            "Nigeria",
            "Rest of West Africa"
        ]
    )
]

visa_table = pd.pivot_table(
    visa_comparison,
    index="group",
    columns="visa",
    values="estimate",
    aggfunc="sum",
    fill_value=0
)

if "TOTAL" in visa_table.columns:

    visa_table = visa_table.drop(
        columns=["TOTAL"]
    )

chi_square, chi_p, chi_dof, expected = (
    stats.chi2_contingency(
        visa_table
    )
)
```

## STATISTICAL TESTS

## HOLT-WINTERS FORECAST

```python
# ============================================================
# 33. HOLT-WINTERS FORECAST
# ============================================================
#
# This is a statistical benchmark, not an official forecast.
# ============================================================

nigeria_annual = (
    annual_complete
    .set_index("year")["nigeria"]
)

holt_winters_model = ExponentialSmoothing(
    nigeria_annual,
    trend="add",
    damped_trend=True,
    initialization_method="estimated"
).fit(
    optimized=True
)

forecast_years = np.arange(
    2026,
    2031
)

forecast_values = (
    holt_winters_model
    .forecast(
        len(forecast_years)
    )
)

forecast = pd.DataFrame(
    {
        "year":
            forecast_years,
        "forecast_nigeria_arrivals":
            forecast_values.values
    }
)

forecast.to_csv(
    TABLE_DIR /
    "holt_winters_forecast_2026_2030.csv",
    index=False
)
```
